```python
df['liked'] = df['liked'].replace({'1':'yes', '0':'no', '2':'yes'})
df['liked'] = df['liked'].fillna(df['liked'].mode()[0])
df['liked'] = df['liked'].map({'no':0, 'yes':1})
```

```python
df['liked'].value_counts()
```

```python
df['category'] = df['category'].replace({'MUsic':'Music','music':'Music','gamingg':'Gaming', 'Ed':'Education',
                                        'COMEDY':'Comedy','Tech ':'Tech'})
```

```python
df['category'].unique()
```

```python
df['timestamp'].isna().sum()
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
```

```python
df['time'] = df['timestamp'].dt.time
df['hour'] = df['timestamp'].dt.hour
df['minute'] = df["timestamp"].dt.minute
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.describe()
```

```python
def video_min_sec(total_seconds):
    # hours = int(total_seconds // 3600)
    minutes = int((total_seconds % 3600) // 60)
    seconds = int(total_seconds % 60)
    return minutes, seconds
    
# Apply and unpack:
df['video_min'], df['video_sec'] = zip(*df['video_duration'].apply(video_min_sec))
```

```python
df.head()
```

```python
df.rename(columns={"video_duration": "video_duration_sec"}, inplace=True)
```

```python
df['is_short_video'] = (df['video_duration_sec'] <= 120).astype(int)
```

```python
# 3. Video duration buckets
df['duration_bracket'] = pd.cut(
    df['video_duration_sec'],
    bins=[0, 300, 900, 1800, 3600, np.inf],
    labels=['shorts_under_5m', 'short_5_15m', 'medium_15_30m', 'long_30_60m', 'extra_long_1h_plus']
)
```

```python
df.shape[0]
```

```python
df['watch_percent'] = (df['watch_percent'] * 100).round(2)
```

```python
df  = df[df['watch_percent'] > 00.00]
```

```python
df = df[df['watch_percent'] <= 100.00]
```

```python
from sklearn.model_selection import train_test_split, StratifiedKFold,cross_val_score
from sklearn.preprocessing import LabelEncoder,OrdinalEncoder,OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline,make_pipeline
```

```python
df = df.sample(100000)
```

```python
df.head()
```

```python
X = df[['video_duration_sec','watch_time','liked','commented','hour','watch_percent','video_min','is_short_video','duration_bracket']]
y = df['category']
```

```python
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(y)
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.3,random_state =42, stratify=y)
```

```python
X_train.shape
```

```python
time_order = ['Morning', 'Afternoon', 'Evening', 'Night']
duration_order = list(X_train['duration_bracket'].cat.categories)

preprocess_pipeline = ColumnTransformer(transformers=[
    ("scaler",MinMaxScaler(),['video_duration_sec','watch_time','watch_percent','hour','video_min']),
    ("ordinal_duration", OrdinalEncoder(categories=[duration_order]), ['duration_bracket']),
],remainder='passthrough')
```

```python
X_train_processed = preprocess_pipeline.fit_transform(X_train)
X_test_processed = preprocess_pipeline.transform(X_test)
```

```python
X_train_processed
```

```python
preprocess_pipeline.get_feature_names_out()
```

```python
feature_names = [col.split('__')[1] for col in preprocess_pipeline.get_feature_names_out()]

X_train_processed = pd.DataFrame(X_train_processed, columns=feature_names)
X_test_processed = pd.DataFrame(X_test_processed, columns=feature_names)
```

```python
X_train_processed
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier,AdaBoostClassifier
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score,classification_report ,confusion_matrix
```

```python
pipelines = {
    "Decision Tree": Pipeline([
        ("classifier",DecisionTreeClassifier(max_depth=20,random_state=42)),
    ]),
    "Random Forest": Pipeline([
        ("classifier", RandomForestClassifier(n_estimators=250,max_depth=16,random_state=42,n_jobs=-1)),
    ]),
    "XGBoost": Pipeline([
        ("classifier", XGBClassifier(random_state=42,n_jobs=-1,)),
    ])
}
```

```python
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
# 2. Train, Evaluate, and Collect Results
results = []

for name, model in pipelines.items():
    # Fit model
    model.fit(X_train_processed, y_train)
    y_pred = model.predict(X_test_processed)

    # Calculate metrics
    acc = accuracy_score(y_test, y_pred)
    cv_mean = np.mean(cross_val_score(model,X_train_processed, y_train, cv=skf,scoring="accuracy",n_jobs=-1))

    results.append({
        "Model": name,
        "Test Accuracy": acc,
        "5-Fold CV Mean": cv_mean,
    })

    print(f"\n{'='*20} {name} {'='*20}")
    print(f"Test Accuracy : {acc:.4f}")
    print(f"CV Mean Acc   : {cv_mean:.4f}\n")
    print("Classification Report:")
    # Set output_dict=False for clean, formatted text output
    print(classification_report(y_test, y_pred))

    # Confusion Matrix Plot
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title(f"{name} - Confusion Matrix")
    plt.tight_layout()
    plt.show()

# 3. Display Overall Comparison Table
summary_df = pd.DataFrame(results).sort_values(by="Test Accuracy", ascending=False)
print("\n" + "="*50)
print("MODEL COMPARISON SUMMARY")
print("="*50)
print(summary_df.to_string(index=False))
```
